```python
# ============================================================
# NAIVE BAYES EXAMPLE 1: Email Spam Filter
# ============================================================
# What is Naive Bayes?
#   It's a simple math-based classifier that asks:
#   "Given these words, what is the PROBABILITY this email is spam?"
#
# Formula (plain English):
#   P(spam | words) = P(words | spam) * P(spam) / P(words)
#
# "Naive" means we ASSUME each word is independent of others.
# That's not always true in real life, but it still works well
# ============================================================

# Step 1: Import the tools we need
from sklearn.naive_bayes import MultinomialNB       # The Naive Bayes model
from sklearn.feature_extraction.text import CountVectorizer  # Turns words into numbers
import numpy as np
```

```python
# ============================================================
# Step 2: Create our DATASET (this is our "training data")
# Each email is a string of words.
# The label tells us: 1 = spam, 0 = not spam
# ============================================================

emails = [
    # --- SPAM emails ---
    "win a free prize now click here",
    "you won free money claim your prize",
    "free offer click now win cash",
    "congratulations you have won free lottery",
    "buy now discount free offer limited time",

    # --- NOT SPAM (ham) emails ---
    "let us meet tomorrow for lunch",
    "can you send me the report",
    "the project deadline is next friday",
    "please review my document and reply",
    "see you at the team meeting tomorrow",
]

# Labels: 1 means SPAM, 0 means NOT SPAM
labels = [1, 1, 1, 1, 1,   # first 5 are spam
          0, 0, 0, 0, 0]   # last 5 are not spam
```

```python
# ============================================================
# Step 3: Convert words into NUMBERS
# Computers can't understand words — they need numbers!
# CountVectorizer counts how many times each word appears.
#
# Example:
#   "free money free" becomes --> {"free": 2, "money": 1}
#   Then turned into a list:  --> [2, 1, 0, 0, ...]
# ============================================================

vectorizer = CountVectorizer()   # Create the converter tool

# fit_transform does TWO things at once:
#   1. "fit"      = learn all unique words from our emails
#   2. "transform" = convert each email into a number list
X = vectorizer.fit_transform(emails)

# X is now a big table of numbers (word counts per email)
print("=== STEP 3: Words converted to numbers ===")
print(f"Number of unique words found: {len(vectorizer.get_feature_names_out())}")
print(f"All unique words: {vectorizer.get_feature_names_out()}")
print()
```

```python
# ============================================================
# Step 4: Create and TRAIN the Naive Bayes model
# Training = showing the model our examples so it can learn
# ============================================================

model = MultinomialNB()    # Create a blank (untrained) model

# Train the model:
#   X      = the word counts for each email
#   labels = the correct answers (spam or not spam)
model.fit(X, labels)

print("=== STEP 4: Model trained! ===")
print()
```

```python
# ============================================================
# Step 5: TEST the model on NEW emails it has never seen
# ============================================================

# These are brand new emails — the model has never seen them
new_emails = [
    "free cash prize win now",         # Looks like spam!
    "can we meet for coffee tomorrow",  # Looks like normal email
    "claim your free lottery ticket",  # Looks like spam!
    "please send me the project files", # Looks like normal email
]

# Convert the new emails into numbers using the SAME vectorizer
# We use transform() only (NOT fit_transform) because we already
# learned the word list during training — don't change it now!
X_new = vectorizer.transform(new_emails)

# Ask the model to make predictions
predictions = model.predict(X_new)

# Get the PROBABILITY of each prediction (more informative!)
probabilities = model.predict_proba(X_new)
```

```python
# ============================================================
# Step 6: Show the results
# ============================================================

print("=== STEP 5 & 6: Predictions on new emails ===")
print()

for i, email in enumerate(new_emails):
    # predictions[i] is either 1 (spam) or 0 (not spam)
    result = "SPAM" if predictions[i] == 1 else "NOT SPAM"

    # probabilities[i][0] = probability of NOT spam
    # probabilities[i][1] = probability of spam
    prob_spam    = probabilities[i][1] * 100   # multiply by 100 to get percent
    prob_notspam = probabilities[i][0] * 100

    print(f"Email: '{email}'")
    print(f"  --> Prediction : {result}")
    print(f"  --> Spam chance: {prob_spam:.1f}%")
    print(f"  --> Ham  chance: {prob_notspam:.1f}%")
    print()
```
